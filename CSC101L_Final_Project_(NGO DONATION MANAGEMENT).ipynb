```python

Companies=[["01","5681983","Grameenphone", 0],
      ["02","5682008","Walton", 500000],
      ["03","5681992","Nestlé Bangladesh Ltd.", 0],
      ["04","5681940","Akij", 700000],
      ["05","5681973","ACI Ltd.", 0],
      ["06","5681978","Aarong" , 200000],
      ["07","5681974","Jamuna Group", 175000],
      ["08","5681988","Sheltech", 0],
      ["09","5681954","ACME Laboratories Ltd.", 0],
      ["10","5681985","Transcom Group", 0]]


def amount_donated():
    print("{0:16}{1:28}{2:40}{3:3}".format("Company No","Company Id","Company Name","Amount"))
    total_donated_amount=0
    for i in range(len(Companies)):
      if Companies[i][3]>=100000:
        print("{0:16}{1:28}{2:40}{3:3}".format(Companies[i][0],Companies[i][1],Companies[i][2],Companies[i][3]))
        total_donated_amount+=Companies[i][3]
    print()
    print(f"Total Donated Amount= {total_donated_amount}tk")



def non_donating_companies():
    print()
    print("The company who haven't donated yet")
    print("{0:16}{1:28}{2:40}{3:3}".format("Company No: ","Company ID: ","Company Name: ","Amount: "))
    for i in range(len(Companies)):
        if Companies[i][3] == 0:
            print("{0:16}{1:28}{2:40}{3:3}".format(Companies[i][0],Companies[i][1],Companies[i][2],Companies[i][3]))


def collect_donation(company_no):
    i = 0
    while i < len(Companies):
        if Companies[i][0] == company_no:
            if Companies[i][3] == 0:
                donation_amount = int(input("Enter the donation amount: "))
                if donation_amount >= 100000:
                    Companies[i][3] += donation_amount
                    print(f"Thank you, {Companies[i][2]}! Successfully recorded your donation of {donation_amount}tk.")
                    break
                else:
                    print(f"Sorry!! {Companies[i][2]} Donation amount should be at least 100000tk.")
                    break
            else:
                print(f"Sorry!! {Companies[i][2]} company has already donated.")
                break
        i += 1
    else:
        print(f"Sorry!! Company with number {company_no} was not found.")



def add_new_company():
    #Declaring local variables
    company_name = input("Enter new company name: ")
    company_id = input("Enter new company ID: ")
    donation_amount = int(input("Enter donation amount: "))
    print()


    if donation_amount >= 100000 or donation_amount == 0:
        #Checking if the company already exists
        for i in range(len(Companies)):
            if Companies[i][2].lower() == company_name.lower():
              if Companies[i][3] == 0:
                print("This company already exists. Adding your donation to the existing entry.")
                Companies[i][3] += donation_amount

              else:
                print("This company already exists.")
              print(f"Your company, {Companies[i][2]}, has donated {Companies[i][3]}tk. Thank you for your contribution.")
              return Companies

        #Adding company
        print("Company not found in the list. Proceeding to add a new company.")
        print()
        Companies.append([str(len(Companies) + 1), company_id, company_name, donation_amount])
        if donation_amount > 0:
            print("Thank you! Your donation has been recorded.")
            print(f"New company added: {company_name}.")
            print(f"You have donated {Companies[-1][3]}tk. Thank you for your contribution.")
        elif donation_amount == 0:
            print("You have successfully registered your company.")
        return Companies

    else:
        print("Sorry, donation amount must be 100,000tk or more.")
        return Companies



def print_menu():
    print("Menu:")
    print("1 -->> View Companies and Total Donations")
    print("2 -->> View Registered Companies That Haven't Donated")
    print("3 -->> Make a donation")
    print("4 -->> Add New Company to the Registry")
    print("5 -->> Print Menu Again")
    print("0 -->> Exit the program")
    print()


print_menu()
while True:
  menu = int(input("Select option (5 to print menu again): "))
  print()
  if menu == 1:
    amount_donated()
    print()
  elif menu == 2:
    non_donating_companies()
    print()
  elif menu == 3:
    company_no = input("Enter the company number: ")
    collect_donation(company_no)
    print()
  elif menu == 4:
    Companies = add_new_company()
    print()
  elif menu == 5:
    print_menu()
  elif menu == 0:
    print("Exiting.")
    break

  else:
    print("Enter a valid input. ")
    print()
```
